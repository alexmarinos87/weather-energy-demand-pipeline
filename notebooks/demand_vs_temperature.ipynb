# Demand by temperature band

This notebook groups retained observations into fixed-width temperature bands. The result is descriptive and does not imply that temperature caused the observed demand.

```python
from forecasting.demo import build_demo_feature_frame
from forecasting.demand_weather_report import (
    DemandWeatherAnalysisConfig,
    build_demand_weather_analysis,
)

features = build_demo_feature_frame()
analysis = build_demand_weather_analysis(
    features,
    config=DemandWeatherAnalysisConfig(temperature_bin_width_c=2.5),
)
```

```python
analysis["temperature_demand_profile"].sort_values(
    ["source_area", "resource_id", "city", "temperature_bin_lower_c"]
)
```
